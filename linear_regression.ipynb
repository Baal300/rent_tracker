# Linear Regression on apartment listings

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/immo_listings.csv")
city_dfs = {
    city: city_df.reset_index(drop=True) for city, city_df in df.groupby("city")    # split data grouped by city into separate dataframes
}

berlin_df = city_dfs["Berlin"]
munich_df = city_dfs["Munich"]
cologne_df = city_dfs["Cologne"]
```

## Berlin Data

```python
plt.figure()
plt.scatter(berlin_df["size"], berlin_df["price"])
plt.show()
```

### Linear Regression

```python
from sklearn.linear_model import LinearRegression 

X = berlin_df["size"].to_numpy().reshape(-1, 1)
y = berlin_df["price"].to_numpy().reshape(-1, 1)

regr = LinearRegression().fit(X,y)
xpts = np.linspace(5, 200, 1000)

plt.figure()
plt.plot(xpts, regr.predict(xpts.reshape(-1,1)))
plt.scatter(berlin_df["size"], berlin_df["price"])
plt.show()
```
