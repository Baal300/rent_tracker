# Linear Regression on apartment listings

Predicting rent prices based on apartment size using linear regression.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/immo_listings_2026-09-28.csv")
city_dfs = {
    city: city_df.reset_index(drop=True) for city, city_df in df.groupby("city")    # split data grouped by city into separate dataframes
}

berlin_df = city_dfs["Berlin"]
munich_df = city_dfs["Munich"]
cologne_df = city_dfs["Cologne"]
```

## Berlin Data

```python
plt.figure()
plt.scatter(berlin_df["size"], berlin_df["price"])
plt.show()
```

### Linear Regression

```python
from sklearn.linear_model import LinearRegression 

X = berlin_df["size"].to_numpy().reshape(-1, 1)
y = berlin_df["price"].to_numpy().reshape(-1, 1)

regr = LinearRegression(fit_intercept=False).fit(X, y)
xpts = np.linspace(5, 200, 1000)

plt.figure()
plt.plot(xpts, regr.predict(xpts.reshape(-1,1)))
plt.scatter(berlin_df["size"], berlin_df["price"])
plt.show()
print(f"Model score: {regr.score(X,y)}")
print(f"Model slope: {regr.coef_}")
print(f"Model intercept: {regr.intercept_}")

input_size = input("Size of apartment in sqm: ")
regr.predict(np.array([[float(input_size)]]))
```
